```python
from pathlib import Path
from zipfile import ZipFile, ZIP_DEFLATED
from google.colab import files

with ZipFile("/content/r0_results.zip", "w", ZIP_DEFLATED) as z:
    for name in ["mlflow.db", "mlruns", "mlartifacts", "logs"]:
        p = Path(name)
        if p.is_file():
            z.write(p)
        elif p.is_dir():
            for f in p.rglob("*"):
                if f.is_file():
                    z.write(f)

files.download("/content/r0_results.zip")
```
